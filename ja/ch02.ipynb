## 11. 先頭からN行を出力
ファイルの先頭N行だけを表示せよ。例えば、N=10として先頭10行を表示せよ。確認にはheadコマンドを用いよ。

```python
def print_head(path: str, n: int):
    with open(path, "r") as f:
        lines = f.readlines()
        for l in lines[:n]:
            print(l, end="")
print_head("../data/popular-names.txt", 10)
```

```python
!head -10 ../data/popular-names.txt
```

## 12. 末尾のN行を出力
ファイルの末尾N行だけを表示せよ。例えば、N=10として末尾10行を表示せよ。確認にはtailコマンドを用いよ。

```python
def print_tail(path: str, n: int):
    with open(path, "r") as f:
        lines = f.readlines()
        for l in lines[-n:]:
            print(l, end="")
print_tail("../data/popular-names.txt", 10)
```

```python
!tail -10 ../data/popular-names.txt
```

## 13. タブをスペースに置換
ファイルの先頭10行に対して、タブ1文字につきスペース1文字に置換して出力せよ。確認にはsedコマンド、trコマンド、もしくはexpandコマンドなどを用いよ。

```python
with open("../data/popular-names.txt", "r") as f:
    lines = f.readlines()
    for l in lines[:10]:
        print(" ".join(l.split("	")), end="")
```

```python
!head -10 ../data/popular-names.txt | sed 's/	/ /g'
```

## 14. 1列目を出力
ファイルの先頭10行に対して、各行の1列目だけを抜き出して表示せよ。確認にはcutコマンドなどを用いよ。

```python
with open("../data/popular-names.txt", "r") as f:
    lines = f.readlines()[:10]
    for l in lines:
        print(l.split("	")[0])
```

```python
!head -10 ../data/popular-names.txt | cut -f -1
```

## 15. ファイルをN分割する
ファイルを行単位でN分割し、別のファイルに格納せよ。例えば、N=10としてファイルを10分割せよ。同様の処理をsplitコマンドで実現せよ。

```python
def split_to_n(path: str, n: int):
    with open(path, "r") as f:
        lines = f.readlines()

    each = []
    per = (len(lines) + 1) // n
    for i in range(0, n):
        each.append(lines[i:i+per])

    for i, e in enumerate(each):
        with open(f"../junk/{i}.txt", "w") as f:
            f.writelines(e)

split_to_n("../data/popular-names.txt", 10)
```

```python
!tldr split
```

```python
%%bash
N=10
lcount=$(wc -l < ../data/popular-names.txt)
per=$(( (lcount + N - 1) / N ))   # 向上取整，保证正好 N 份
split -l "$per" ../data/popular-names.txt ../junk/
```

## 16. ランダムに各行を並び替える
ファイルを行単位でランダムに並び替えよ（注意: 各行の内容は変更せずに並び替えよ）。同様の処理をshufコマンドで実現せよ。

```python
import random
def shuf(path: str):
    with open(path) as f:
        lines = f.readlines()
        random.shuffle(lines)
    with open("../junk/shuffled.txt", "w") as f:
        f.writelines(lines)

shuf("../data/popular-names.txt")
```
